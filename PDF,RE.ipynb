```python
text = "The phone number of the agent is 0900-78601 phone. Call soon."
import re
pattern = "phone"
re.search(pattern,text)
```

```python
my_match = re.search(pattern, text)
```

```python
my_match.span()
```

```python
my_match.start()
```

```python
all_matches = re.findall("phone", text) 
len(all_matches)
```

```python
for match in re.finditer("phone",text):
    print(match.span())
```

```python
text
```

```python
pattern = r'\d\d\d\d-\d\d\d\d\d'
phone_number = re.search(pattern,text)
phone_number
```

```python
phone_number.group()
```

```python
pattern= r"\d{4}-\d{5}"
```

```python
re.search(pattern,text)
```

```python
pattern= r"(\d{4})-(\d{5})" #to make group
match=re.search(pattern,text)
match.group(1) # it will find the group one on index one 
```

```python
re.search(r"man|woman","this man was here ")
```

```python
re.findall(r".at","The cat in the hat sat ") #this is wildcard, it will find words with at  
```

```python
re.findall(r"..at","The cat in the hat sat splat") #add 2 dots (multiple wildcard) to grab 2 letters before at. 
```

```python
re.findall(r"^\d", '1 this ends with a number 2') #^ starts with , # $ ends with
```

```python
phrase = "there are 3 numbers 34 inside 5 this sentence"
```

```python
re.findall(r"[^\d]",phrase) #it will exclude insde this square bracket
```

```python
re.findall(r"[^\d]+",phrase) #we will use plus to give everything except digits.
```

```python
test_phrase = "This is a string! but it has punctuation. How to remove it?" 
```

```python
my_list = re.findall(r"[^!.?]+", test_phrase) # + means one or more time
```

```python
' '.join(my_list)
```

```python
text = "only find the hyphen-words. Where are the long-ish dash-words?"
re.findall(r'[\w]+-[\w]+',text) #grab any alphanumeric. both number and alphabets #here it will find hyphen words
```
